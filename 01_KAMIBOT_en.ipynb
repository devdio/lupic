# Finding the Kamibot connection port

```python
import pykamilab

# 1) Search for available ports
pykamilab.list_ports()
```

# Connect to and initialize Kamibot

```python
# 2) Connect - response_timeout is recommended (the cell auto-stops if the firmware does not respond)
from pykamilab import KamibotPi
bot = KamibotPi('COM69', response_timeout=3.0)
bot.init()

```

```python
# 3) Move forward 5 cm
bot.move_forward_unit(5, "-l")
bot.delay(1)
bot.stop()

# # 4) Shut down (keep the kernel alive)
# bot.disconnect()
```

# Changing the LED color

```python
# Set the RGB values directly (each 0-255)
bot.turn_led(255, 0, 0)
bot.delay(1)   # red
bot.turn_led(0, 255, 0)
bot.delay(1)   # green
bot.turn_led(0, 0, 255)
bot.delay(1)   # blue
```

# Sounding the buzzer

```python
bot.melody(45, 0.5)
```

```python
C4 = 60   # Do  (C)
D4 = 62   # Re  (D)
E4 = 64   # Mi  (E)
F4 = 65   # Fa  (F)
G4 = 67   # Sol (G)
A4 = 69   # La  (A)
B4 = 71   # Si  (B)
C5 = 72   # high Do (C)

twinkle = [
    (C4, 0.5), (C4, 0.5), (G4, 0.5), (G4, 0.5),
    (A4, 0.5), (A4, 0.5), (G4, 1.0),
    (F4, 0.5), (F4, 0.5), (E4, 0.5), (E4, 0.5),
    (D4, 0.5), (D4, 0.5), (C4, 1.0),
]

for note, length in twinkle:
    bot.melody(note, length)
```
